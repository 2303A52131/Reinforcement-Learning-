```python
import gymnasium as gym

def random_agent(env_name, episodes=5):
    env = gym.make(env_name)

    print("\n===================================")
    print("Environment:", env_name)
    print("===================================")

    episode_rewards = []
    for episode in range(episodes):

        state, info = env.reset()

        total_reward = 0
        steps = 0
        terminated = False
        truncated = False
        while not (terminated or truncated):
            action = env.action_space.sample()
            next_state, reward, terminated, truncated, info = env.step(action)
            total_reward += reward
            steps += 1
            state = next_state
        average_reward = total_reward / steps if steps > 0 else 0

        episode_rewards.append(total_reward)
        print("\nEpisode", episode + 1)
        print("Total Reward:", total_reward)
        print("Average Reward:", round(average_reward, 2))
        print("Steps:", steps)

    overall_average = sum(episode_rewards) / len(episode_rewards)

    print("\n-----------------------------------")
    print("Overall Average Reward:",
          round(overall_average, 2))
    print("-----------------------------------")
    env.close()
random_agent("CliffWalking-v1", episodes=5)
random_agent("Taxi-v4", episodes=5)
```
