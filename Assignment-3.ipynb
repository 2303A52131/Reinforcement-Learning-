```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Given parameters
M = 3
N_play = 10
epsilon = 0.2

# Part B - Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# Given reward patterns
rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of reward position
reward_index = {'A': 0, 'B': 0, 'C': 0}

# Fixed seed for reproducible output
random.seed(10)

# Run for 10 plays
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Find highest Q-value
    max_Q = max(Q.values())

    # Machines having highest Q-value
    best_machines = [m for m in machines if Q[m] == max_Q]

    # Epsilon-greedy decision
    if r < epsilon:

        # Exploration: exclude highest-Q machine
        other_machines = [
            m for m in machines if m not in best_machines
        ]

        if other_machines:
            selected_machine = random.choice(other_machines)
        else:
            selected_machine = random.choice(machines)

        decision = "Exploration"

    else:

        # Exploitation
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    # Get reward from given pattern
    index = reward_index[selected_machine]

    # Repeat pattern if necessary
    if index >= len(rewards[selected_machine]):
        index = 0

    reward = rewards[selected_machine][index]
    reward_index[selected_machine] += 1

    # Update N
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    # Display play information
    print("\nPlay", play)
    print("Random number:", round(r, 3))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)
    print("Reward:", reward)
    print("Updated Q-value:", round(Q[selected_machine], 2))


# Final results
print("\n======================================")
print("           FINAL RESULTS")
print("======================================")

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        machine,
        "\t", N[machine],
        "\t", R[machine],
        "\t", round(Q[machine], 2)
    )

# Total reward
total_reward = sum(R.values())

print("\nTotal Reward =", total_reward)

# Best machine
best_machine = max(Q, key=Q.get)

print("Best Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 2))
```
