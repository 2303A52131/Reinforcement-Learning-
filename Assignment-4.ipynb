```python
import math

# -----------------------------------------
# Part A - Machine Initialization
# -----------------------------------------

machines = ['A', 'B', 'C']

# -----------------------------------------
# Part B - Initialize N, R, Q and UCB
# -----------------------------------------

N = [0, 0, 0]              # Number of times each machine is selected
R = [0, 0, 0]              # Total reward of each machine
Q = [0.0, 0.0, 0.0]        # Estimated average reward
UCB = [0.0, 0.0, 0.0]

total_plays = 10
total_reward = 0

# -----------------------------------------
# Part C, D and E
# UCB calculation, selection,
# reward collection and Q update
# -----------------------------------------

for t in range(1, total_plays + 1):

    # Calculate UCB for each machine
    for i in range(len(machines)):

        if N[i] == 0:
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt(
                (2 * math.log(t)) / N[i]
            )

    # Select machine with highest UCB
    selected = max(
        range(len(machines)),
        key=lambda i: UCB[i]
    )

    print("\nPlay", t)
    print("Selected Machine:", machines[selected])

    # Get reward from user
    reward = float(input("Enter reward: "))

    # Update number of selections
    N[selected] += 1

    # Update total reward for selected machine
    R[selected] += reward

    # Incremental sample-average update
    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    # Update total reward
    total_reward += reward

    print("Reward:", reward)
    print("Updated Q:", round(Q[selected], 4))


# -----------------------------------------
# Part F - Final Results
# -----------------------------------------

print("\n========================================")
print("             FINAL RESULTS")
print("========================================")

print("\nMachine\tN\tR\tUCB\t\tQ")

# Calculate final UCB values
for i in range(len(machines)):

    if N[i] == 0:
        UCB[i] = float('inf')
    else:
        UCB[i] = Q[i] + math.sqrt(
            (2 * math.log(total_plays)) / N[i]
        )

    print(
        f"{machines[i]}\t"
        f"{N[i]}\t"
        f"{R[i]:.2f}\t"
        f"{UCB[i]:.4f}\t\t"
        f"{Q[i]:.4f}"
    )

print("\nTotal Reward Received:", total_reward)
```
